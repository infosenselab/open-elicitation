```python
# ---------------- Imports ----------------

import os
import json

import yaml
```

```python
# ---------------- Inputs ----------------

INPUT_FILENAME = "ibm-rank-30k-samp2/meta-llama/Llama-3.1-8B-Instruct/20260912t001205-llm-consistency.jsonl"

```

```python
# ---------------- Config ----------------

with open("../../config/config.yaml", "r") as f:
    config = yaml.safe_load(f)

proj_store = config["paths"]["proj-store"]

INPUT_PATH = os.path.join(proj_store, "experiments", "llm-consistency", INPUT_FILENAME)
```

```python
unique_arguments = {}

with open(INPUT_PATH, "r", encoding="utf-8") as f:
    for line in f:
        row = json.loads(line)
        unique_arguments.setdefault(row["argument_id"], row["argument"])

for argument_id, argument in unique_arguments.items():
    print(f"{argument_id}: {argument}")

print(f"\nTotal unique arguments: {len(unique_arguments)}")
```
